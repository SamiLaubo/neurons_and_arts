## Prune wrong data

Not all neurons can be downloaded due to a wrongfull position of the neuron from the initial table or a decompression fault. We simply skip these neurons and remove their folders.

````python
from pathlib import Path
import shutil
path = Path("../emimesh/results/cells_lowres/")
i = 0
for cell_type in path.glob("*/"):
    for cell in cell_type.glob("*/"):
        if not (cell / "meshes/mesh.png").exists():
            print(cell)
            shutil.rmtree(cell) # Remove folder
            i += 1
    
print(f"Removed {i} cells without meshes.")
````

## Remove raw files
Remove raw segementation files to reduce memory.

````python
from pathlib import Path
path = Path("../emimesh/results/cells_lowres/")
i = 0
for vti in path.rglob("*.vti"):
    vti.unlink()
    i += 1
print(f'Removed {i} .vti files')
````
