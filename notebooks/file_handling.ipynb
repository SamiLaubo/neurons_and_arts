# Handle generated data

## Prune wrong data

Not all neurons can be downloaded due to a wrongfull position of the neuron from the initial table or a decompression fault. We simply skip these neurons and remove their folders.

````python
from pathlib import Path
import shutil
path = Path("../emimesh/results/cells/")
removed_idx = []
for cell_type in path.glob("*/"):
    for cell in cell_type.glob("*/"):
        print(cell)
        if not (cell / "meshes/mesh.png").exists():
            # Remove folder
            removed_idx.append(int(cell.name.split("_")[-1]))
            shutil.rmtree(cell)
    
print(f"Removed {len(removed_idx)} cells without meshes: {removed_idx}")
````

## Remove raw files

````python
from pathlib import Path
path = Path("../emimesh/results/cells/")
removed_idx = []
for cell_type in path.glob("*/"):
    for cell in cell_type.glob("*/"):
        if (cell / "raw/raw.vti").exists():
           (cell / "raw/raw.vti").unlink() # Delete 
````
